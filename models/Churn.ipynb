```python
df = pd.read_csv('customer_churn_dataset.csv')
```

```python
df.columns
```

```python
df = df.drop(['CustomerID', 'Last Interaction'], axis=1)
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])
df['Subscription Type'] = le.fit_transform(df['Subscription Type'])
df['Contract Length'] = le.fit_transform(df['Contract Length'])
```

```python
X = df.drop('Churn', axis=1)
```

```python
X.head()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans
```

```python
inertia = []
K = range(1, 11)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.plot(K, inertia, 'bx-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method to find optimal k')
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)
labels4 = kmeans.fit_predict(X_scaled)
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)
labels4 = kmeans.fit_predict(X_scaled)
df['Cluster']=labels4
```

```python
y_pred=kmeans.predict(X_scaled)
```

```python
y_pred
```

```python
from sklearn.metrics import silhouette_score
```

```python
silhouette_k3 = silhouette_score(X_scaled, labels4)
```

```python
print("Silhouette Score for k=4:", silhouette_k3)
```

```python
# Cluster profiling
cluster_profile = df.groupby('Churn').mean()
print(cluster_profile)

# Compare churn rates across clusters
import seaborn as sns
sns.countplot(x='Cluster', hue='Churn', data=df)
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Reduce data to 2D with PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Add PCA components to dataframe
df['PCA1'] = X_pca[:, 0]
df['PCA2'] = X_pca[:, 1]

# Plot clusters
plt.figure(figsize=(8,6))
sns.scatterplot(x='PCA1', y='PCA2', hue='Cluster',
                palette='coolwarm', data=df, s=60, alpha=0.7)

plt.title("Customer Clusters (k=4) visualized with PCA", fontsize=14)
plt.legend(title="Cluster")
plt.show()
```
